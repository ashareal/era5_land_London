```python
import cdsapi
from pathlib import Path

out_dir = Path("data/era5land")
out_dir.mkdir(parents=True, exist_ok=True)

client = cdsapi.Client()

months = [(2001, 12)] + [(y, m) for y in range(2002, 2007) for m in range(1, 13)]

for year, month in months:
    target = out_dir / f"era5land_london_{year}_{month:02d}.nc"
    if target.exists():
        continue  # already downloaded
    request = {
        "variable": ["2m_temperature", "2m_dewpoint_temperature"],
        "year": str(year),
        "month": f"{month:02d}",
        "day": [f"{d:02d}" for d in range(1, 32)],
        "time": [f"{h:02d}:00" for h in range(24)],
        "area": [51.70, -0.52, 51.28, 0.34],  # N, W, S, E
        "data_format": "netcdf",
        "download_format": "unarchived",
    }
    print("requesting", target.name)
    client.retrieve("reanalysis-era5-land", request, str(target))
```
